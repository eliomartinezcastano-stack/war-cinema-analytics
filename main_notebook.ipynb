```python
def get_conflict_intensity_by_decade(df, min_year=1930, max_year=2024):
    decade_list = []

    for i in range(len(df)):
        start = df.loc[i, "start_year"]
        end = df.loc[i, "end_year"]

        if pd.isna(end):
            end = max_year

        start = int(start)
        end = int(end)

        for year in range(start, end + 1):
            if min_year <= year <= max_year:
                decade = (year // 10) * 10
                decade_list.append(decade)

    df_intensity = (
        pd.Series(decade_list)
        .value_counts()
        .sort_index()
        .reset_index()
    )

    df_intensity.columns = ["decade", "intensity"]
    df_intensity["intensity"] = df_intensity["intensity"].astype(int)

    return df_intensity
```

```python
df_intensity = get_conflict_intensity_by_decade(df_wars_final)
df_intensity
```

MERGE

```python
h1 = pd.merge(war_movies_by_decade, df_intensity, on='decade', how='left')
h1["intensity"] = h1["intensity"].fillna(0).astype(int)

h1.sort_values('decade')
```

```python
df_intensity_h2 = df_intensity.copy()
df_intensity_h2["decade_plus_one"] = df_intensity_h2['decade'] + 10
```

```python
h2 = pd.merge(war_movies_by_decade, df_intensity_h2[["decade_plus_one", "intensity"]], left_on='decade', right_on="decade_plus_one", how="left").drop(columns=["decade_plus_one"])
```

```python
h2["intensity"] = h2["intensity"].fillna(0).astype(int)
```

```python
h2 = h2.rename(columns={"intensity" : "intensity_prev_decade"})
```

```python
final_compare = pd.merge(
    h1,
    h2[["decade", "intensity_prev_decade"]],
    on="decade",
    how="left"
)

final_compare = final_compare.sort_values("decade")
final_compare
```
